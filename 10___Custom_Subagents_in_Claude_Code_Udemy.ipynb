```python
'''

Hi there 👋 , Very practical question now :


    [ SHOW in VS CODE ]
   "OK , I want to BUILD a custom sub-agent ..
    WHERE does Claude actually expect this file ?"


Lets answer it step by step 🚀


------------------------------------------------------------
🔹 The Storage Convention
------------------------------------------------------------

👉 Every custom sub-agent lives as a single .md file inside :
      .claude / agents /  <agent-name> .md


   So the structure looks like :

      [diagram]

           .claude /                      ← Claude's config folder
              └── agents /                ← sub-folder for ALL custom agents
                     ├── security-auditor.md
                     ├── code-reviewer.md
                     └── blog-writer.md





------------------------------------------------------------
🔹 The 2 TYPES of Custom Sub-agents ( based on location )
------------------------------------------------------------

      [diagram]

                          Custom Sub-agents
                                  |
                   ---------------+---------------
                   |                              |
            User / Project-               Project-level
            level                          ( also called : repo-level )

            Inside  HOME directory         Inside  PROJECT directory
            of .claude                     of .claude

            ie .. ~/.claude/agents/        ie .. ./.claude/agents/
                       *.md                          *.md



👉 USER-LEVEL  ( system-wide )

   Path  ->>  ~/.claude/agents/<name>.md
   Available to ALL projects on your machine ✅




👉 PROJECT-LEVEL  ( scoped to ONE project )

   Path  ->>  <your-project>/.claude/agents/<name>.md

   ie ..
        /my-sales-project/
            ├── .claude/
            │      └── agents/
            │             ├── sales-data-cleaner.md
            │             └── eda-runner.md
            ├── src/
            └── ...

   Available ONLY inside THIS project ✅
 
 




------------------------------------------------------------
🔹 What does ONE custom agent file actually look like ?
------------------------------------------------------------

👉 A Custom sub-agent is a MARKDOWN file with 2 clear parts :


      [diagram]

           sub-agent .md file
                  |
         ---------+----------
         |                  |
         |                  |
   YAML frontmatter   Main body of agent
                       ( agent description )


👉 YAML frontmatter   ->>   the CONFIG of your agent
                            ( name , description , tools , model , etc. )

👉 Main body          ->>   detailed instructions for the sub-agent
                            ie ..  HOW it should perform its task ✨


Example layout :

      ---
      name: security-auditor
      description: Use this agent for company-grade security audits .
      tools: Read, Grep, Glob
      model: opus
      ---

      You are a senior application security engineer .
      Always follow our internal security checklist :
         - All APIs must validate JWT
         - No raw SQL queries
         - No hardcoded secrets
      ...



In the next section , we will deep-dive into the YAML FRONTMATTER part :
& the KEYS that actually configure your custom sub-agent 🗝️

'''
```

# 5.. [Live Demo] : The 7 Keys of a Custom Sub-agent  🗝️
        First Show VS CODE , then PPT , then back to VS CODE


```python
'''

Hi there 👋 , We Remember the structure of a custom sub-agent file ?

      ---
      <YAML frontmatter>          ← the KEYS live here
      ---

      <main body of sub-agent>    ← how to perform the task


Inside the YAML frontmatter , there are some KEYS you fill in ..
These keys define WHAT the sub-agent is & WHAT it can do ✅


------------------------------------------------------------
🔹 The 4 CORE Keys ( every custom sub-agent needs these )
------------------------------------------------------------

      [diagram]

                       sub-agent .md
                            |
                ------------+-------------
                |                        |
         YAML frontmatter            Main body
                                     ( "how to perform task" )
                |
                |
        ┌───────────────┐
        |  name         |   ←  sub-agent name
        |  description  |   ←  description of sub-agent
        |  tools        |   ←  set of tools agent can use
        |  model        |   ←  which Claude model to use
        └───────────────┘


Apart from these 4 keys , you can give MANY MORE keys to sub-agents :

      ┌───────────────┐
      |  skills       |   ←  what skills this sub-agent has access to
      |  hooks        |   ←  pre / post hooks for this sub-agent
      |  memory       |   ←  memory behaviour
      |  effort       |   ←  effort level ( reasoning depth )
      |  color        |   ←  UI color tag
      └───────────────┘


So in total ->>  4 core  +  5 optional = 9 keys you can play with 🔥




Lets unlock each ONE of them ✨


------------------------------------------------------------
🗝️  KEY 1   →   name
------------------------------------------------------------

👉 The unique identifier of your sub-agent ..
👉 Used when you call it manually  ( eg : @sales-analysis-agent )



------------------------------------------------------------
🗝️  KEY 2   →   description
------------------------------------------------------------

👉 👉 Explains WHEN Claude should use this agent .
👉 Claude READS this to decide if this agent fits the user's prompt ✨

    Example :

   description: Use this agent for Sales Data Analysis ,
                KPI dashboards , revenue trends ,
                SQL analytics , and business insights .


------------------------------------------------------------
🗝️  KEY 3   →   tools
------------------------------------------------------------

👉 The set of TOOLS this sub-agent is allowed to use ..

Common tools :
   Read , Write , Edit , Grep , Glob , Bash , WebSearch , WebFetch

Examples :
   👉  Research agent     ->>   tools: Read, WebSearch, WebFetch
   👉  Code Writer agent  ->>   tools: Read, Write, Edit, Grep, Glob




------------------------------------------------------------
🗝️  KEY 4   →   model
------------------------------------------------------------
👉 Which Claude model should this sub-agent use ?


💰 example :
   👉  Research agent   ->>   opus      ( deep reasoning )
   👉  code-reviewer      ->>   sonnet    ( balanced )





------------------------------------------------------------
🔹 Now lets unlock the BONUS keys
------------------------------------------------------------

Apart from the 4 core keys , you can give many more keys to sub-agents ..
These are OPTIONAL  ->>  but very useful in real projects ✨


------------------------------------------------------------
🗝️  KEY 5  ( bonus )   →   skills
------------------------------------------------------------

👉 What about various skills access to sub-agents ?

   👉 sales-dashboard-agent
          ->> skills: streamlit , charts

   👉 sql-analysis-agent
          ->> skills: sql , database
          
    👉 csv-cleaning-agent
          ->> skills: pandas , preprocessing
   

Now when this agent runs ..
it already KNOWS the best practices encoded in those skills ✅




------------------------------------------------------------
🗝️  KEY 6  ( bonus )   →   hooks
------------------------------------------------------------

👉 What about hooks for sub-agents ?

👉 Hooks  =  Safety Protocols for your Project that runs BEFORE / DURING / AFTER agent runs ..

Use-cases :
    📌 Sales Data Analysis Use-cases :
       👉 PRE-hook
              ->> validate sales CSV before analysis
       👉 PRE-hook
              ->> create backup of dataset
       👉 POST-hook
              ->> auto-run data quality checks


------------------------------------------------------------
🗝️  KEY 7  ( bonus )   →   memory
------------------------------------------------------------

👉 This key will Control how the sub-agent uses memory across runs ..
Options :
   👉 none
          ->> fresh Sales Analysis every time
   👉 session
          ->> remembers KPIs & dataset context
               during same VS Code session
   👉 persistent
          ->> remembers business rules ,
               custom KPIs & dashboard preferences
               across sessions





------------------------------------------------------------
🗝️  KEY 8  ( bonus )   →   effort
------------------------------------------------------------

👉 Effort level  ->>  how HARD should the agent think before responding ?

Options :
   👉  low       ->>  fast , minimal reasoning
   👉  medium    ->>  balanced
   👉  high      ->>  deep reasoning  ( for audits , architecture , planning )

Examples :
   👉  csv-column-renamer   ->> effort: low
   👉  dashboard-generator  ->> effort: medium






------------------------------------------------------------
🗝️  KEY 9  ( bonus )   →   color
------------------------------------------------------------

👉 Sets the COLOR of this agent in the Claude Code UI ✨

Common options :   blue , green , red , yellow , purple , orange ..

Example :
   👉 green
          ->> safe read-only agents
               ( explore-agent )
   👉 red
          ->> destructive agents
               ( file-cleanup-agent )
   👉 yellow
          ->> dashboard / visualization agents
          
          
Now while working on large Sales Analytics projects :
   👉 you can instantly identify agent types














------------------------------------------------------------
🔹 Full Example  →  All Keys Together
------------------------------------------------------------

A complete custom sub-agent .md file might look like this :

       ---
       name: security-auditor
       description: Use this agent for company-grade security audits .
                    Triggers on words like "audit" , "vulnerability check" ,
                    "security scan" . Follow our internal security
                    checklist for every finding .
       tools: Read, Grep, Glob
       model: opus
       skills:
         - company-security-checklist
       hooks:
         post:
           - command: "echo 'Audit complete' >> audit.log"
       memory: none
       effort: high
       color: red
       ---

       You are a senior application security engineer at <your company> .
       Your ONLY job is to find security vulnerabilities .

       Always follow our internal security checklist :
         - All APIs must validate JWT
         - No raw SQL queries  ( must use ORM )
         - No hardcoded secrets ( must come from Vault )
         - Every endpoint must have rate limiting
         - Cite exact file + line number for every finding

       End with a severity-ranked summary table .


See how READABLE this is ?
👉 Frontmatter  ->>  config
👉 Main body    ->>  the detailed instructions ( how to perform the task )

Both pieces TOGETHER  ->>  give birth to your custom AI specialist 🔥


------------------------------------------------------------
🔹 Quick Summary of All Keys
------------------------------------------------------------

      🗝️  Key            |  What it controls                          |  Required ?
      ------------------|--------------------------------------------|-------------
      name              |  unique identifier of the agent             |  ✅ Yes
      description       |  WHEN Claude should auto-route to it        |  ✅ Yes
      tools             |  what tools the agent can use               |  ✅ Yes
      model             |  opus / sonnet / haiku / inherit            |  ✅ Yes
      skills            |  attached how-to playbooks                  |  optional
      hooks             |  pre / post actions around agent runs       |  optional
      memory            |  none / session / persistent                |  optional
      effort            |  low / medium / high                        |  optional
      color             |  UI color tag                                |  optional


Master these keys  ->>  and you can build ANY kind of specialist
sub-agent for ANY project 🔥


And in the next section , we will see  ->>  HOW to actually
CALL these custom sub-agents inside Claude Code 📞

'''
```

# 6.. How to CALL your Custom Sub-agents  📞
        First Show VS CODE , then PPT , then back to VS CODE


```python
'''

Hi there 👋 , So far we have learnt :

   👉 What custom sub-agents are
   👉 Why they BEAT built-ins for company-specific work
   👉 Where they are stored ( user-level vs project-level )
   👉 The 4 core keys + 5 bonus keys

Now the BIG question :

👉 How do you actually CALL / TRIGGER these custom sub-agents
   inside Claude Code ?


There are 2 ways .. lets see both 🚀


------------------------------------------------------------
🔹 The 2 Ways to Call a Custom Sub-agent
------------------------------------------------------------

      [diagram]

                       How to Call Custom Sub-agents
                                  |
                  ----------------+-----------------
                  |                                  |
            Automatically                       Manually
        (Claude decides ,                  (Programmer triggers
         based on description)              sub-agent explicitly)



------------------------------------------------------------
1)  AUTOMATICALLY   ( the smart way )
------------------------------------------------------------

👉 Claude can AUTOMATICALLY call sub-agent by looking into the
   DESCRIPTION of the sub-agent  ->>  as per the work you asked .

Remember the   description:   field in the YAML frontmatter ?

      ---
      name: security-auditor
      description: Use this agent for company-grade security audits ..
      ...
      ---


Behind the scenes :
   👉 Claude scans ALL the   description:   fields of your custom agents
   👉 Matches your prompt's intent to the right description
   👉 Auto-triggers the matching sub-agent ✨


Example :

   You type in Claude Code :
      "Do a security audit on my repo .."

   Claude reads descriptions  ->>  finds  security-auditor  matches
   →  auto-triggers it  ✅

   Sub-agent runs in its OWN isolated context window ,
   does the audit using YOUR company guidelines ,
   returns clean findings to the main agent ✅


💡 BIG INSIGHT :
   👉 The QUALITY of your   description:   field decides how often
      Claude routes to this agent correctly ..
   👉 Write descriptions like "Use this when ..." with TRIGGER KEYWORDS 🔥


------------------------------------------------------------
2)  MANUALLY   ( the explicit way )
------------------------------------------------------------

👉 You can ALSO call a custom sub-agent MANUALLY ..
👉 ie - YOU explicitly trigger the sub-agent ..


There are 2 popular patterns for MANUAL triggering :


PATTERN  A   ->>   @agent-name  /  natural-language explicit call
   👉 You just mention the agent by name in your prompt ..

   Example :
      "@security-auditor please audit this repo .."
      "@code-reviewer review my last 3 commits .."

   OR natural language :
      "Use the security-auditor sub-agent on this codebase .."
      "Trigger my blog-post-writer to draft this article .."

   Claude sees the explicit call  →  triggers that SPECIFIC agent ✅


PATTERN  B   ->>   Trigger from a CUSTOM SLASH COMMAND 🔥
   👉 This is a SUPER powerful pattern most beginners miss ..

   The idea :
      You can create your OWN workflow ( a custom slash command )
      where you MENTION  ->>  when this custom sub-agent should be called .

   ie  inside a custom slash command file ,
   you say  ->>  "call this sub-agent when this command is invoked"


   Example  ->>   /test  command :

      .claude / commands / test.md

         ---
         description: Run our testing workflow for the project .
         ---

         Run unit tests for the most recent changes in this project .
         👉 For this task , trigger our  unit-test-generator  sub-agent .
         👉 Make sure all new functions are covered .


   Now when you type :
       /test

   Behind the scenes :
       👉 Claude opens the test slash command
       👉 Sees the instruction to call  unit-test-generator
       👉 Triggers that sub-agent automatically ✅


   So the chain is :
       Slash Command   →   triggers  →   Custom Sub-agent  🔁


   In short  ->>
   You have mentioned in /test feature's slash command :
   "call this sub-agent when I want to perform testing .."


💡 This is the senior-engineer pattern :
   👉 Slash commands  =  the trigger / shortcut
   👉 Custom sub-agents  =  the specialist worker behind it
   Together  ->>  fully automated , one-command workflows 🔥


------------------------------------------------------------
🔹 Side-by-side Comparison
------------------------------------------------------------

      Way               |  How it works                |  Best for
      ------------------|------------------------------|------------------------
      Automatic         |  Claude reads description    |  Day-to-day usage
                        |  & auto-picks the agent      |  when prompts are clear
      ----------------- |------------------------------|------------------------
      Manual (@name)    |  You explicitly @mention     |  Debugging / specific
                        |  the agent in your prompt    |  agent control
      ----------------- |------------------------------|------------------------
      Manual via        |  Slash command's body tells  |  Reusable team
      Slash Command 🔥  |  Claude to call the agent    |  workflows / shortcuts



------------------------------------------------------------
🔹 Hands-on Flow ( end-to-end )
------------------------------------------------------------

Lets put it all together in ONE real flow 🚀

✅ STEP 1  -  Create your custom agent file
   👉 .claude/agents/security-auditor.md   ( project-level )
   👉 Add the YAML frontmatter + main body  ( from Section 4 )


✅ STEP 2  -  Launch Claude Code
   👉 Open VSCode  ->>  open terminal  ->>  run :
         claude


✅ STEP 3  -  Trigger AUTOMATICALLY
   👉 Type a natural prompt :
         "Do a security audit on this repo .."

   👉 Claude reads custom-agent descriptions  →  auto-routes to
      security-auditor  ✅


✅ STEP 4  -  Trigger MANUALLY  ( @ mention )
   👉 Type :
         "@security-auditor scan only the auth module .."

   👉 Claude triggers that EXACT agent ✅


✅ STEP 5  -  Trigger MANUALLY  ( via slash command ) 🔥
   👉 Create  .claude/commands/audit.md  with body :
         Run our company security checklist on the codebase .
         👉 Use the  security-auditor  sub-agent for this task .

   👉 In Claude Code , type :
         /audit

   👉 Slash command fires  →  it triggers the security-auditor agent ✅


✅ STEP 6  -  Verify with the Observer  ( from previous section )
   👉 Open the agents-observe URL in browser
   👉 Confirm that  security-auditor  was triggered ✅
   👉 See its ISOLATED context window in action 👀


------------------------------------------------------------
🔹 Pro Tips Before We Wrap
------------------------------------------------------------

   👉 1.  Keep   name   short & lowercase ( easy to @mention )
   👉 2.  Invest HEAVILY in   description   ( it powers auto-routing )
   👉 3.  Start with Manual calls ( predictable )  ->>  graduate to Automatic
   👉 4.  Combine Slash Commands + Custom Sub-agents  ->>  killer workflows 🔥
   👉 5.  Version-control project-level agents in Git ->> team-wide superpowers
   👉 6.  Use the observer to debug which agent fires for which prompt 👀


------------------------------------------------------------
🔹 Final Mental Model
------------------------------------------------------------

      Built-in agents      ->>  come pre-installed by Claude
                                ( generic knowledge )

      Custom agents        ->>  YOU design them ( frontmatter + body )
                                ( YOUR company's specialised knowledge )

      Trigger them by      ->>  Automatically  ( via description matching )
                                OR Manually    ( @name  OR via slash command )


And just like that  ->>   YOU now have your OWN team of AI specialists
working for you 24/7  ->>   exactly the way you designed them 🔥


🎉 Congratulations !
You have officially completed the Custom Sub-agents section .
In the next section we will continue building on this foundation
with more powerful Claude Code features ..

See you there 👋

'''
```
